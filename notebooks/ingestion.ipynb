```python
# [Cell 1] Import thư viện và viết hàm Schema Discovery
import requests
import json

def extract_schema(data):
    """
    Hàm đệ quy để quét toàn bộ dữ liệu thô và trích xuất định dạng (Data Types).
    """
    if isinstance(data, dict):
        return {key: extract_schema(value) for key, value in data.items()}
    elif isinstance(data, list):
        # Lấy định dạng của phần tử đầu tiên làm đại diện cho toàn bộ list
        if len(data) > 0:
            return [extract_schema(data[0])]
        return ["empty_list"]
    else:
        return type(data).__name__

print("Đã khởi tạo Engine quét cấu trúc dữ liệu.")
```

```python
# [Cell 2] BINANCE: Exchange Information & Trading Rules Metadata
binance_url = "https://api.binance.com/api/v3/exchangeInfo?symbol=BTCUSDT"
response = requests.get(binance_url)

if response.status_code == 200:
    raw_data = response.json()
    
    print("========== BINANCE RAW METADATA ==========")
    print(json.dumps(raw_data, indent=2))
    
    print("\n========== BINANCE RETURN FORMAT (SCHEMA) ==========")
    schema = extract_schema(raw_data)
    print(json.dumps(schema, indent=2))
else:
    print(f"Lỗi truy xuất Binance: {response.status_code}")
```

```python
import asyncio
import aiohttp
import time
import pandas as pd
from datetime import datetime, timezone

# 1. Định nghĩa cấu hình các nguồn (URLs và logic chuẩn hóa riêng)
SOURCES_CONFIG = {
    "Binance": {
        "url": "https://api.binance.com/api/v3/ticker/price?symbol=BTCUSDT",
        # Binance trả JSON: {"symbol": "BTCUSDT", "price": "90000.00"}
        "parser": lambda data: float(data['price'])
    },
    "Coinbase": {
        "url": "https://api.coinbase.com/v2/prices/BTC-USD/spot",
        # Coinbase trả JSON: {"data": {"amount": "90000.00", ...}}
        "parser": lambda data: float(data['data']['amount'])
    },
    "Kraken": {
        "url": "https://api.kraken.com/0/public/Ticker?pair=XBTUSD",
        # Kraken lồng ghép sâu: {"result": {"XXBTZUSD": {"c": ["90000.00", ...], ...}}}
        "parser": lambda data: float(list(data['result'].values())[0]['c'][0])
    },
    "CoinGecko": {
        "url": "https://api.coingecko.com/api/v3/simple/price?ids=bitcoin&vs_currencies=usd",
        # CoinGecko: {"bitcoin": {"usd": 90000.00}}
        "parser": lambda data: float(data['bitcoin']['usd'])
    }
}

async def fetch_and_normalize(session, source_name, config, anchor_time):
    """
    Task bất đồng bộ: Gọi API, ghi nhận độ trễ và chuẩn hóa dữ liệu.
    """
    try:
        # Giới hạn thời gian chờ để không làm treo toàn bộ batch
        async with session.get(config['url'], timeout=5) as response:
            if response.status == 200:
                raw_data = await response.json()
                
                # Tính thời gian phản hồi (từ mốc neo đến khi nhận xong body)
                latency_ms = round((time.time() - anchor_time) * 1000, 2)
                
                # Gọi hàm parser tương ứng để chuẩn hóa ra số thực
                normalized_price = config['parser'](raw_data)
                
                return {
                    "Source": source_name,
                    "Status": "Success",
                    "Price (USD)": normalized_price,
                    "Latency (ms)": latency_ms,
                }
            else:
                return {"Source": source_name, "Status": f"HTTP {response.status}", "Price (USD)": None, "Latency (ms)": None}
    except Exception as e:
        return {"Source": source_name, "Status": f"Error: {str(e)}", "Price (USD)": None, "Latency (ms)": None}

async def sync_and_compare():
    """
    Hàm Orchestrator: Gom nhóm các task và phân tích kết quả.
    """
    print(f"Bắt đầu đồng bộ tại UTC: {datetime.now(timezone.utc)}")
    
    async with aiohttp.ClientSession() as session:
        # BƯỚC 1: Lập mốc neo thời gian ngay trước khi bắn request
        anchor_time = time.time()
        
        # BƯỚC 2: Tạo danh sách các task bất đồng bộ
        tasks = [
            fetch_and_normalize(session, name, config, anchor_time)
            for name, config in SOURCES_CONFIG.items()
        ]
        
        # Kích hoạt tất cả các task CHẠY SONG SONG cùng một thời điểm
        results = await asyncio.gather(*tasks)
        
        # BƯỚC 3 & 4: Đưa vào cấu trúc bảng và tính toán so khớp
        df = pd.DataFrame(results)
        df_success = df[df["Status"] == "Success"].copy()
        
        print("\n=== KẾT QUẢ ĐỒNG BỘ DỮ LIỆU CÙNG THỜI ĐIỂM ===")
        display(df)
        
        if not df_success.empty:
            max_price = df_success['Price (USD)'].max()
            min_price = df_success['Price (USD)'].min()
            spread_usd = max_price - min_price
            spread_pct = (spread_usd / min_price) * 100
            
            fastest_source = df_success.loc[df_success['Latency (ms)'].idxmin(), 'Source']
            
            print("\n=== PHÂN TÍCH SO KHỚP (RECONCILIATION) ===")
            print(f"-> Mức giá Cao nhất : {max_price:,.2f} USD")
            print(f"-> Mức giá Thấp nhất: {min_price:,.2f} USD")
            print(f"-> Độ lệch (Spread) : {spread_usd:,.2f} USD ({spread_pct:.4f}%) - Yếu tố phát hiện cơ hội Arbitrage.")
            print(f"-> Nguồn phản hồi nhanh nhất mạng lưới: {fastest_source}")

# Thực thi (Trên Jupyter/Google Colab thì chạy trực tiếp, nếu file python thì dùng asyncio.run(sync_and_compare()))
await sync_and_compare()
```
